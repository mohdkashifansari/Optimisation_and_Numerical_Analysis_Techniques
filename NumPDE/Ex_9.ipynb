```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81
h0 = 1.0
lam = np.sqrt(g * h0)

x = np.linspace(0, 10, 400)

def f(x):
    return np.exp(-x**2)

t = 2.0

y = 0.5 * (f(x - lam*t) + f(x + lam*t))
v = (1/(2*lam)) * (f(x - lam*t) - f(x + lam*t))

# plot
plt.figure()
plt.plot(x, y, label='y (height perturbation)')
plt.plot(x, v, label='v (velocity)')
plt.title(f'Traveling waves at t={t}')
plt.legend()
plt.xlabel('x')
plt.grid()

plt.show()
```
