```python
import numpy as np
import matplotlib.pyplot as plt

alpha = 1.0
N = 50
h = 1.0 / N

gamma = 0.4
tau = gamma * h**2

T = 0.005
M = int(T / tau)

x = np.linspace(0, 1, N + 1)

u = np.zeros(N + 1)
u[x >= 0.5] = 1.0

u[0] = 0.0
u[-1] = 1.0

for k in range(M):

    u_new = u.copy()

    u_new[1:-1] = (
        u[1:-1]
        + alpha * gamma *
        (u[2:] - 2*u[1:-1] + u[:-2])
    )

    u_new[0] = 0.0
    u_new[-1] = 1.0

    u = u_new

plt.plot(x, u)
plt.xlabel("x")
plt.ylabel("u")
plt.title(f"Explicit scheme, gamma = {gamma}")
plt.grid(True)
plt.show()
```
