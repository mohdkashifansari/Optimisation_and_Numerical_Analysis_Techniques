```python
import numpy as np
import time
a, b = -2, 2
dx = 0.002
dt = 0.001
c = 1
T_e = 2

x = np.arange(a, b, dx)
u = np.exp(-6 * x**2)

start = time.time()

t = 0
while t <= T_e:

    u_shift = np.roll(u, 1)
    
    u = u - c * dt/dx * (u - u_shift)
    
    t += dt

end = time.time()
print("runtime:", end - start)
```
