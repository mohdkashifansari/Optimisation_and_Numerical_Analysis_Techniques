```python
import numpy as np
import matplotlib.pyplot as plt

# parameters
t = 1.0                     # time
x = np.linspace(-1, 2, 1000)  # spatial grid

# define the solution
u = np.zeros_like(x)
u[x > t] = 1.0
mask = (x > 0) & (x <= t)
u[mask] = np.sqrt(x[mask] / t)

# plot
plt.figure()
plt.plot(x, u)
plt.xlabel("x")
plt.ylabel("u(t,x)")
plt.title(f"Entropy solution at t = {t}")
plt.grid(True)
plt.show()
```
